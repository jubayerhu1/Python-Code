## Performance Metrics in classification 

```python
from sklearn.metrics import (accuracy_score,confusion_matrix, precision_score, f1_score, fbeta_score,recall_score)
```

```python
# Example true labels and predictions
y_true = [0, 1, 1, 0, 1, 0, 1, 1]
y_pred = [0, 1, 0, 0, 1, 0, 1, 0]
```

```python
# Acuracy 
accuracy = accuracy_score(y_true, y_pred)
print("Accuracy :",accuracy)
```

```python
# Confusion Matrix
cm = confusion_matrix(y_true, y_pred)
print("Confusion Matrix:\n",cm)
```

```python
# Recall 
recall = recall_score(y_true, y_pred)
print('Recall :', recall)
```

```python
# F1 score
f1 = f1_score(y_true, y_pred)
print("F1 Score :",f1)
```

```python
# F-beta score (example: beta = 0.5 gives more weight to precision)

fbeta = fbeta_score(y_true, y_pred, beta=0.5)
print("F- beta score (beta=0.5): ",fbeta)
```
