Question: Write a program to print numbers from 10 to 15 using a while loop

```python
#======================================================================
# Reverse Each word in a sentence 
# Question : Write a program to reverse each word in the sentence "Hello Jubayer" using a while loop
# Expected Output : olleh reyabuj
#======================================================================


sentence  = 'Hello Jubayer'
words = sentence.split()
print(words)

for word in words:
    i = len(word) -1

    while i>=0:
        print(word[i], end=" ")
        i -=1
    print(end=" ")
```
