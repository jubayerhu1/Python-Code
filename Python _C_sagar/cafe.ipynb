```python
# Menu dictionary
menu = {'Pizza': 20, 'Pasta': 40, 'Burger': 10, 'coffee': 3}

print("Welcome to Python Diner!\nPizza: $20 | Pasta: $40 | Burger: $10 | Coffee: $3\n" + "-"*35)

total = 0
item = input("Enter your item: ")

if item in menu:
    total += menu[item]
    print(f"✅ {item} added!")
else:
    print("❌ Item not available.")

another = input("Would you like anything else? (yes/no): ")

if another.lower() == "yes":
    item2 = input("Enter 2nd item: ")
    if item2 in menu:
        total += menu[item2]
        print(f"✅ {item2} added!")
    else:
        print("❌ Item not available.")

print("-" * 35)
print(f"Your total bill is: ${total}")
```
