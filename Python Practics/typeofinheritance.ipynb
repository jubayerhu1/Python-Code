Types of Inheritance in OOP

- single inheritance
- Multilevel inheritance
- Hirerachical inheritance 
- Multiple inheritance 


```python
'''Types of Inheritance in OOP

- single inheritance
- Multilevel inheritance
- Hirerachical inheritance 
- Multiple inheritance '''

# Multilevel Inheritance 

# Grandfather

class Product:
    def review(self):
        print("product customer review")

# Father
class phone(Product):
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying a phone..")

#Child 
class smartphone(phone):
    pass

s = smartphone(200, "Apple", 3)
s.buy()
s.review()
```

```python
# Multilevel Inheritance


#Gradfather
class Product:
  def review(self):
    print("Product customer review")


#Father
class Phone(Product):
  def __init__(self, price, brand, camera):
    print("Inside Phone constructor")
    self.__price = price
    self.brand = brand
    self.camera = camera


  def buy(self):
    print("Buying a phone")


#child
class SmartPhone(Phone):
  pass



S  = SmartPhone(200000, "Samsung", 3)
S.buy()
S.review()
```

```python
# Hierarchical 

# parent 
class phone:
    def __init__(self, price, brand, camera):
        print("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying a phone ")

# child
class smartphone(phone):
    pass

#child
class featurephone(phone):
    pass

s = smartphone(100, "apple", 3)
s.buy()
```

# Polymorphism in OOP 

Polymorphism refers to having multiple forms.Polymorphism is a programming term that refers to the use of the same function name, but with different signatures.


```python
class shape:
    def area(self, redius):
        print("This is 1st")
        return 3.14 * redius ** 2

    def area(self, l,b):
        print("This is 2nd ")
        return l*b


s = shape()
s.area(2,5)
```

```python
class shape:
    def area(self, redius):
        print("This is 1st ")
        return 3.14 * redius ** 2

    def area(self, l,b):
        print("This is 2nd ")
        return l*b

s = shape()
s.area(2)
```
