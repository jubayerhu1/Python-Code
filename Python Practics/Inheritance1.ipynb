Inheritance in OOP 

Inheritance allows us to define a class that inherits all the methods and properites from another class. Parent class is the class being inherited from, also called base class that inherits from another class, also called derived class

DRY - Don't Repeat yourself

```python
class user: # parent class
    def __init__(self):
        self.name = "Jubayer"
        self.gender = "Male"
    
    def login(self):
        print("login done -parent ")

    def register(self):
        print("register done ")

```

```python
class student(user): # child class
    def __init__(self):
        self.rollno =12

    def enroll(self):
        print("Enrolled the course")


class instructor(user):
    def __init__(self):
        self.idno = 120

    def crated(self):
        print("Created the course- instructor ")

    def reply(self):
        print("Replay instructor")
```

```python
s = student()
s.login()
s.enroll()

i = instructor()
i.crated()
i.login()
i.register()
i.reply()
```
