```python
# Countdown

i = 5 
while i > 0:
    print(i)
    i-=1
```

```python
# Ask user until valid input
# using isnumer

num = ""
while not num.isnumeric():
    num = input("Enter any value : ")
    
print("Number accepted : ",num)
```

```python
# Ask user until valid input 
# using isnumeric()

num = ""

while not num.isnumeric():
    num = input("Enter any value : ")

    if not num.isnumeric():
        print("Try agin ! ")
print("Number accepted ", num)
```
