```python
# Ask user until valid input
# using isnumer

num = ""
while not num.isnumeric():
    num = input("Enter any value : ")
    
print("Number accepted : ",num)
```

```python
# Ask user until valid input 
# using isnumeric()

num = ""

while not num.isnumeric():
    num = input("Enter any value : ")

    if not num.isnumeric():
        print("Try agin ! ")
print("Number accepted ", num)
```

```python
# password system (Advanced )

password = ""
attemps = 0 

while password != "admin123" and attemps < 3:
    password = input("Enter your password : ")
    attemps +=1

    if password == "admin123" :
        print("Login Successful ")
    else:
        print("Wrong password try again , Attempts count : ", attemps)
    if attemps ==3:
        print(" 3 Attemps expired ")
```
