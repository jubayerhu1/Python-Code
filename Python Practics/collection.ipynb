Collection of class object in OOP 

```python
class person: 
    def __init__(self, name, country):
        self.name = name
        self.country = country
```

```python
p1 = person("jubayer","usa")
p2 = person("jihan", "UK")
p3 = person("sayem", "ud")
```

```python
l = [ p1, p2, p3]


for i in l:
    #print(i)
    print(i.name, i.country)
```

```python
class Customer:
    def __init__(self,name, gender,address):
        self.name = name
        self.gender = gender
        self.address = address

    def print_infor(self):
        print(f"Name :{self.name}, Gender :{self.gender}, Address:{self.address.city}, {self.address.pin},State : {self.address.state}")

class Address:
    def __init__(self, city,pin, state):
        self.city = city
        self.pin = pin
        self.state = state
```

```python
adds = Address("Daka", 1230, "BD")

cust = Customer ("Jubayer", "Male",adds)
```

```python
cust.print_infor()
```
