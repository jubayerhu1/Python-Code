# Tuple 
- Tuple are used to store multiple items in single variable.
- Tupe is one of 4 built-in data types in python used to store collections of data, The other 3 are List, Set, and Dictionary, all with different qualities and usage.
- A Tupe is a collection which is ordered and unchangeable. 

```python
t = ('apple',)
print(t)
t = ('apple', 'banana', 'cherry')
print(t)
```

```python
# Tuples can also be created without the parentheses:

tuple = "apple", "banna", "cherry"
print(tuple)
```

```python
# Tuple Items 
# Tupe items are ordered, unchangeable, and allow duplicate values.
# Tupe items are indexed, the first item has index[0] , the second item has index [1] etc.

# Ordered 
# When we say that tuples are orderd, it means that the items have a defined order, and that order will not change.

#Unchangeable
# Tuples are unchangeable, meaning that we cannot - change, add or remove -- items after the tupe has been created.

# Allow Duplicates 
# Since tupes are indexed, they can have items with the same value:
```

```python
Tuple = ("apple", "banana", "cherry", "apple")
print(Tuple)
```

```python
# Tupe lengt 
# To determine how many items a tuple has, use the -- len() function:
print(len(Tuple))
```

```python
# Create Tuple with one item
# To create a tuple with only one item, you have to add a comma, after the item, otherwise python will not recognize it as a tuple.

# One item tuple , remember the comma:
t = ('apple',)
print(type(t))
t = ('apple')
print(type(t))
```

```python
# create an empty tuple:

# To crate an empty tuple, use round brackets with no content 
t = ()
print(type(t))
```

```python
# Tuple items _ Data Types:

# Tuple items can be of any data type:
# String , int and boolean data type:
tuple1 = ("apple", "banana", "cherry")
tuple2 = (1, 5, 7, 9, 3)
tuple3 = (True, False, False)
```

```python
# A tuple with string, integers and boolean values:
tuple1 = ("abc", 34, True, 40, "male")
print(tuple1)
```

```python
# From python's perspective, tuples are defined as objects with the data type 'tuple'

print(type(tuple1))
```

Python Collections (Arrays)

There are four collection data types in the python programming language:

- List is a collection which is ordered and changeable. Allows duplicate members.
- Tuple is a collection which is ordered and unchangeable. Allows duplicate members.
- Set is a collection which is unordered, unchangeable* , and unindexed. No duplicate members.
- Dictionary is a collection which is ordered ** and changeable . No duplicate members.
