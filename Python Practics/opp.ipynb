```python
class AtmMachine:

    def __init__(self):
        print("Start program")
        self.pin = ""
        self.balance = 0

        self.menu()


#===========================================

    def menu(self):
        user_input = input(
            """
            Hi welcome to Atm Machine 

            1. press 1 to crate pin
            2. press 2 to change pin
            3. press 3 to check balance 
            4. press 4 to withdraw 
            5 . Antything to exit the program
            
            """
        )

        if user_input == "1":
            #crate pin
            self.create_pin()
            print("====")
            
            #change pin
        elif user_input == "2":
            self.Change_pin()
        
        # check balance 
        elif user_input =="3":
            self.check_balance()

        elif user_input == "4":
            self.withdraw_balance()
          
        else:
            print('coming code')

#=============================
#
#============================
    def create_pin(self):

        user_pin= input("Enter your pin : ")
        self.pin = user_pin

        user_balance = float(input("Enter your balance : "))
        self.balance = user_balance 

        print("Succeesfully pin crate ")

        self.menu()
#========================================================
    def Change_pin(self):
        old_pin= input("Enter your old pin :")
        if old_pin == self.pin:
            new_pin = input("Enter your new pin :")
            print("pin change successfully ")
            self.menu()
        else:
            print("your pin is incorrent , please try ")
            self.menu()
 #=====================================================   
    def check_balance(self):
        user_pin = input("Enter your pin")

        if user_pin == self.pin:
            print("Your balance is : ",self.balance)
        else:
            print("Your pin is incorrect, please try again")
        
        self.menu()

#=============================
#
#============================
    def withdraw_balance(self):
        user_pin = input("Enter your pin ")

        if user_pin ==self.pin:
            amount = float(input("Enter the amount : "))
            if amount<= self.balance:
                self.balance -= amount
                print(f"You have withdraw {amount}. Your new balance is {self.balance}")
            else:
                print("Insufficient balance !")
        else:
            print("You pin is incorrect, please try again !")

        self.menu()

            
        
        


   



atm = AtmMachine()
```
