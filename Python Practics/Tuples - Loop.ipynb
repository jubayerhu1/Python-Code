Python - Loop Tuples
Loop Through a Tuple
You can loop through the tuple items by using a - for - loop

```python
# Iterate through the items and print the values:

tp =("apple", "banana", "cherry")

for x in tp:
    print(x)
```

```python
# Loop Throuh the Index Numbers :
# You can also loop through the tuple items by referring to their index number:

# use the range() and len() functions to create a suitable iterable.
```

```python
tp = ("apple", "banana", "cherry")
for i in range(len(tp)):
    print(tp[i])
```

Using a While Loop 
Use the len() function to determine the length of the tuple, then start at 0 and loop your way through the tuple items by referring to their indexes.

Remember to increase the index by 1 after each iteration.

```python
tp = ("apple", "banana", "cherry")

i = 0 
while i< len(tp):
    print(tp[i])
    i = i+1
```

Join Two Tuples
 To join two or more tuples you can use the + operator:

 

```python
t1= ('a','b','c')
t2 = (1,2,3)

t3 = t1+ t2
print(t3)
```

```python
# Multiple Tuples:
# If you want to multiple the content of a tuple a given number of times, you can use the * operator:
fruits = ('apple', 'banana', 'cherry')

mt= fruits *2
print(mt)
mt= fruits *3
print(mt)
```

```python
# count() Retrun the number of times the values 5 appears in the tuple:
# The count() method returns the number of times a specified value appears in the tuple.

thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5,5)

x = thistuple.count(5)

print(x)
```
