**** Looping using list comprehension
List comprehension offers the shortest sysntax for looping through lists:
List comprehensio offers a shorter syntax when you want to create a new list based on the values of an existing list.
Example:
Base on a list of furits, you want a new list , containg only the fruits with the letter "a" in the name:
without list comprehension you will have to write a - for - statement with a conditional test inside :


```python
l = ['apple', 'banana', 'cherry', 'apple', 'mango'] 

newl= []
count = 0 

for x in l:
    if "a" in x:
        newl.append(x)
        count +=1
print(newl)
print(count)
    

```

```python
# with list comprehension you can do all that with only one line of code:

new= [ x for x in l if 'a' in x]
new
```

newlist =[expression for item in iterable if condition == true]



```python
# Condition : The conditionis like a filter that only accepts the items that evaluate to True

# Only accept items that are not "apple"

newlist = [ x for x in l if x != 'apple']
print(newlist)
# the codition if x != 'apple' will return True for all elements other then "apple ", making the new lsit contain all
# fruit or l except "apple
#  the condition is optional and can be omitted"


newl = [ x for x in l]
print(newl)
```

```python
# iterable  the iterable can be any iterable object, like a list , tupe, set etc.

# you can use the range() function to create an iterable

newlist = [x for x in range(10)]
print(newlist)
```

```python
# Same example , but with a condition :

# Accept only numbers lower than 5:
newlist = [ x for x in range(10) if x <5]
print(newlist)
```

```python
# The expression is the current item in the iteration , but it is also the outcome, which you can manipulate
# before it ends up like a list item in the new list

newlist = [x.upper() for x in l]
print(newlist)
```

```python
# You can set the outcome to whatever you like :

# set all values in the new list to 'hello'

newlist = ['hello' for x in l]
print(newlist)
```

```python
# The expression can also contain conditions , not like a filter , but as a way to manipulate the outcome:

# return 'orange' instead of 'banana'

l = ['apple', 'banana', 'cherry', 'apple', 'mango'] 

newlist = [ x if x != 'banana' else 'orange' for x in l]
print(newlist)
```

Python - Join list
There are several ways to join or concatenate , two or more list in python
one of the easiest ways are by using the + operator



```python
l = ['apple', 'banana', 'cherry', 'apple', 'mango'] 
l2= [10,20,30,40]
l3 = l+l2
print(l3)
```

```python
# Another way to join two list is by appending all the items from l2 into l1 , one by one.
l = ['apple', 'banana', 'cherry', 'apple', 'mango'] 
l2= [10,20,30,40]


for x in l2:
    l.append(x)
print(l)
```

```python
# You can use the extend() method , where the purpose is to add elements from one list to another lsit 
l = ['apple', 'banana', 'cherry', 'apple', 'mango'] 
l2= [10,20,30,40]

l.extend(l2)
print(l)
```
