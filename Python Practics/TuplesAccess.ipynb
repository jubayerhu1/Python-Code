# Python - Access Tuple Items

### Access Tuple Items
- You can access tuple items by referring to the index number, inside square brackets:

```python
tp = ('apple', 'banana', 'cherry')
print(tp)
print(tp[1])
```

```python
# Negative Indexing 
# Negative indexing means start from the end.
# -1 refers to the last item, -2 refers to the second last item etc.

print(tp[-1])
```

```python
# Range of Indexes 
# You can specify a range of indexes by specifying where to start and where to end the range.
# When specifing a range, the return value will be a new tupe with the specificed items.



tp = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(tp[2:5])
```

```python
# By leaving out the end value, the range will go on the the end of the tuple:
print(tp[2:])
```

```python
# Range of Negative Indexes
# Specify negative indexes if you want to start the search from the end of the tuple:
tp = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")

print(tp[-4:-1])
```

```python
# Check if Item Exists
# To determind if a specified item is present in a tuple use in the keyword"


tp = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")

if "apple" in tp:
    print("Yes, 'apple' is in the fruits tuple")
```
