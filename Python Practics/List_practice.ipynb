```python
# import the random package here "import random"

import random

def generate_random_list():

    list=[]
    randomlength = random.randint(1, 100)

    for i in range(randomlength):
        list.append(randomlength)
        i+=1
    return list

r =generate_random_list()
lastOne = r[len(r)-1]
print(lastOne)
```
