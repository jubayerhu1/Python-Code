# Python Update Tuples
<p> Tuples are unchangeable, meaning that you cannot change, add , or remove items once the tuple is created.
But there are some workarounds.
# Change Tupe values 
# Once a tuple is created, you cannot change its valuse. Tuples are unchangeable ,
 or immutable as it also called
 But there is a workaround. You can convert the tuple into a list, change ,the list
 , and convert the list back into a tuple.</p>

```python
# Convert the tuple into a list to be able to change it :
x = ("apple", "banana", "cherry")

y = list(x)
y[1]="kiwi"
x =tuple(y)
print(x)
```

<p> Since tuples are immutable, they do not have a built -in append() method. but 
there are other ways add item to a tuple.
Convert into a list : Just like the workaround for changing a tupele , you can convert it into a list, add your items(s)
and convert it back into a tuple.</p>

```python
tp = ("apple", "banana", "cherry")
print(tp)
y = list(tp)
print(y)
y.append('orange')
tp = tuple(y)

print(tp)
```

Add tuple to a tuple . You are allowed to add tuples, so if you want to add one item, (or many), create a new tupe with the item(s), and add it to the existing tuple.

```python
tp = ("apple", "banana", "cherry")

y = ("mango",)
tp += y
print(tp)
```

# Remove the items 
You can not remove the items in a tuple.
Tuples are unchangeable, so you cannot remove the items from it, but you can use the same workaround as we used for changing and adding tuple items:


```python
tp = ("apple", "banana", "cherry")
y = list(tp)
y.remove('apple')
tp=tuple(y)
print(tp)
```

```python
# Del , you can delete the tuple completely:

# The del keyword can delete the tuple completely:

tp = ("apple", "banana", "cherry")
del tp
print(tp) # this will raise an error because the tuple no longer exists
```
