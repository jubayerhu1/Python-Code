```python
# pip install openai
import os
from openai import OpenAI

client = OpenAI(
    api_key=os.environ["EURI_API_KEY"],
    base_url="https://api.euron.one/api/v1/euri",
)

resp = client.chat.completions.create(
    model="euri-auto",
    messages=[
        {"role": "system", "content": "You are a helpful tutor."},
        {"role": "user", "content": "Explain recursion with a simple Python example."},
    ],
    max_tokens=2048,
)
print(resp.choices[0].message.content.strip())
print(resp.model, resp.usage.total_tokens, "tokens")  # euri-auto reports the alias it chose
```
