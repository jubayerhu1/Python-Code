# Python - Access set items

Access items 
You cannot access items in a set by referrig to an index or key.
 But you can loop through the set items using  a -- for - - , or ask if  a specified value is present in a set , by using the in keyword.

```python
set = {"apple", "banana", "cherry"}

for x in set:
  print(x)
```

```python
# Check if "banana " is present in the set 

set = {"apple", "banana", "cherry"}

print("banana" in set)
```

```python
# check if "banana" is present in the set:

set = {'a', 'b', "c"}

print("a" not in set)
```

Change items
