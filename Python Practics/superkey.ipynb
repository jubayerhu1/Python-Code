```python

class phoen :
    def __init__(self, price, brand, camera):
        self.price = price
        self.brand = brand
        self.camera = camera

#=============================

class smartphone(phone):
    def __init__(self, price, brand, camera, os, ram):
        print("Inside smartphone constructor - child class __")
        super().__init__(price, brand, camera)
        self.os = os
        self.ram = ram
        
        print(os,ram)
```
