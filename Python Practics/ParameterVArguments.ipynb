Types of Arguments :
- Default Arggument
- Positional Argument
- Keyword Argument



```python
#- Default Arggument
def power(a,b): # paramater
    return a**b


res = power(4,3) # Argument 
print(res)
```

```python
def power(a=2,b=3): # paramater | defautl parameter
    return a**b


res = power(3) # Argument | if you missed any argument it will work 
print(res)

# many application has defult argument ,set up 
```

```python
# Positiona Argument :
def power(a=2, b=3):
    return a**b

res = power(8,4)
print(res)
```

```python
# key word Argument 

def power(a=2, b=3):
    return a**b

res = power(a=8,b=4)
print(res)
```

```python
# '*arg and **kwargs
# *arg and **kwargs are special keywords that are used to pass the variable lenth of arguments to a function.

# allows us to pass a variable number of non keyword argument to a function.
```

```python
# *args :  allows us to pass a variable number of non keyword argument to a function.

def multiply (a, b):
    return a*b 

res = multiply(2,3)
print(res)
```

```python
def multiply (a, b,c):
    return a*b 

res = multiply(2,3,3)
print(res)
```

```python
def mul(*args): # it is tupe *ju any name but * is required .
    product = 1
    for i in args:
        product =product * i
    return product

#res= mul(1,2,3,5,6,6,4,7,8)
res= mul(1,2,2,3)
print(res)
```
