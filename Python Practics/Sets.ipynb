# Set items
- Set items are unordered , unchangeable, and do not allow duplicate values.
# Unordered
- Unordered means that the items in a set do not have a defined order.
- Set items can appear in a different order every time you use them, and cannot be referred to by index or key.
# Unchangeable
- Set items are unchangeable, meaning that we cannot change the items after the the set has been creted.
- Once a set is created, you cannot change its items, but you can remove items and add new items.

# Duplicate Not Allowed
- Sets cannot have two items with the same value.

```python
s = {"apple", "banana", "cherry", "apple"}
print(s)
print(len(s))
```

```python
# The values True and 1 are considerd the same value in sets, and are treated as duplicates:
# Example: True and 1 is considered the same value:
s= {"apple", "banana", "cherry", "apple", True, 1,2}
print(s)
```

Note : The values False and ) are cnsidered the same value in sets, and are treated as duplicates;



```python
s= {"apple", "banana", "cherry", "apple", False, True, 0}
print(s)
```

```python
# Get the length of a set 
# To determine how many items a set has , use the len() function.
s= {"apple", "banana", "cherry", "apple"}
print(len(s))
```

```python
set1 = {"apple", "banana", "cherry"}
set2 = {1, 5, 7, 9, 3}
set3 = {True, False, False}
```

```python
# A set with strings, integers and boolean values:
set = {"abc", 34, True, 40, "male"}
print(set)
```

The set() Constructor
It is also possible to use the-- set() --constructor to make a set.

```python
s = set(("abc", 34, True, 40, "male")) # note the double round - brackets
print(s)
```
