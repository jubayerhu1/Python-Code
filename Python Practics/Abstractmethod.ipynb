# Abstraction in OOP 
- Abstruction is a fundamental concept in python programming that allows us to simplify complex concept concepts and focus on the essential details. It involves hiding unnecessary details and exposing only the relvent information to the users.

```python
# Abstract Base class (ABC)
from abc import ABC, abstractmethod
# senior 
class server(ABC):
    def database(self):
        print("Connect ot data base")
    
    
    @abstractmethod
    def securityPass(self):
        print("security pass")

class webserverAPP(server):

    def web_login(self):
        print("This is web login -")


    def securityPass(self):
        print("Pass")


    
```

```python
m = webserverAPP()
```

```python
m.web_login()
```
