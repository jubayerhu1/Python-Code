# Add items
Once a set is created, you cannot change its items, but you can add new items.

To add one items to set use the add() method

```python
# Add an items to a set , using the add() method:
set = {"apple", "banana", "cherry"}

set.add('orange')
print(set)
```

# Add sets 
To add items from another set into the current set, use the update() method.


```python
# Add elements from tropical into thisset:
set = {"apple", "banana", "cherry"}
to= {"pineapple", "mango", "papaya"}

set.update(to)
print(set)
```

Add any iterable 
The objectg in the update() method does not have to be aset , it can be any iterable object (tuples, lists, dictiorary)



```python
set = {"apple", "banana", "cherry"}
to= {"pineapple", "mango", "papaya"}

set.update(to)
print(set)
```
